```python
import pandas as pd

import os
import glob
import re
```

```python
file_list = glob.glob('/data2/haochun/Sennet/ISH/E2/coordinates_v3/*.txt')
```

```python
file_list
```

```python
import pandas as pd

channel_gene_map = {
    'R1_Ch1': 'Sfn',
    'R1_Ch2': 'Cdkn2a',
    'R1_Ch3': 'Krt10',
    'R1_Ch4': 'Cdkn1a',
    'R2_Ch1': 'Sbsn',
    'R2_Ch2': 'Dmkn',
    'R2_Ch3': 'Eepd1',
    'R2_Ch4': 'Lor',
    'R3_Ch1': 'Serpine1',
    'R3_Ch2': 'Scn7a',
    'R3_Ch3': 'Abcg1',
    'R3_Ch4': 'Dab2'
}

all_dfs = []

for file_path in file_list:
    for key in channel_gene_map.keys():
        if key in file_path:
            gene=channel_gene_map[key]
    
    print(file_path,gene)
    df = pd.read_csv(file_path, sep='\t', header=0)
    
    df['target'] = gene

    all_dfs.append(df)

merged_df = pd.concat(all_dfs, axis=0, ignore_index=True)

print(merged_df.head())
print(f"Merged dataframe shape: {merged_df.shape}")

```

```python
merged_df.columns = ['x','y','target']
```

```python
merged_df.to_csv('/data2/haochun/Sennet/ISH/E2/spots_decoded_all_stitched_globalcoords_v1.txt', index=False)
```
